```python
import pandas as pd
```

## A. Position and Label-based Slicing

```python
cars = pd.read_csv('cars.csv')
print(cars.shape) # check
```

```python
cols = cars.columns.tolist()
cols # check
```

```python
cars_6_to_10 = cars.iloc[5:10]
cars_6_to_10 # check if valid
```

```python
cars_6_to_10_select = cars_6_to_10.loc[:, ['Model', 'mpg', 'cyl', 'hp', 'gear']]
cars_6_to_10_select # check if valid
```

```python
# combine all requests

print(cars.shape)
print(cols)
display(cars_6_to_10)
display(cars_6_to_10_select)
```

## B. Model Lookup 

```python
toyota = cars[cars['Model']=='Toyota Corolla']
toyota # check
```

```python
pontiac = cars.loc[cars['Model']=='Pontiac Firebird', ['Model', 'mpg', 'hp', 'wt']]
pontiac # check
```

```python
# combine requests

display(toyota)
display(pontiac)
```

## C. Multi-Model Subsetting

```python
selected_cars = cars.loc[(cars['Model']=='Datsun 710')|(cars['Model']=='Lotus Europa')|(cars['Model']=='Ferrari Dino'), ['Model','mpg','cyl','hp','gear']]
```

```python
# combine requests

print(selected_cars.shape)
display(selected_cars)
```
